#1. tf.constant() vs. tf.Variable()

##A constant is a tensor whose value you don't modify through TensorFlow's variable-assignment operations.

```python
import tensorflow as tf

x = tf.constant(10)

print(x)
```

##A tf.Variable is designed to hold a value that can be updated. Neural networks commonly use variables for their weights and biases.

```python
w = tf.Variable(10.0)

print(w)
```

| Feature | `tf.constant()` | `tf.Variable()` |
| :--- | :--- | :--- |
| **Purpose** | Hold a tensor value | Hold an updatable value |
| **Update with `.assign()`** | No | Yes |
| **Used for trainable weights** | Not normally | Yes |
| **Can participate in calculations** | Yes | Yes |

# 3. What is a derivative?

Suppose we have this mathematical function:

$$y = x^2$$

If $x = 3$, then:

$$y = 3^2 = 9$$

The derivative tells us how quickly the output changes when the input changes.

For this function:

$$\frac{dy}{dx} = 2x$$

When $x = 3$:

$$\frac{dy}{dx} = 2(3) = 6$$

The derivative is 6.

**Why is this important?** During neural-network training, gradients tell an optimizer how to adjust weights to reduce prediction errors.

##Our first gradient with TensorFlow

```python
import tensorflow as tf

x = tf.Variable([3.0,6.0,9.0])

with tf.GradientTape() as tape:
    y = x ** 2

gradient = tape.gradient(y, x)

print("y =", y.numpy())
print("Gradient =", gradient.numpy())
```

## 4. A slightly more complex function

Now try:
$$y = 3x^2 + 2x + 1$$

Its derivative is:
$$\frac{dy}{dx} = 6x + 2$$

At $x = 2$:
$$\frac{dy}{dx} = 6(2) + 2 = 14$$

Let's calculate this using TensorFlow.

```python
x = tf.Variable(2.0)

with tf.GradientTape() as tape:
    y = 3 * x**2 + 2 * x + 1

gradient = tape.gradient(y, x)

print("Output:", y.numpy())
print("Gradient:", gradient.numpy())
```

#Gradients for multiple variables

```python
w = tf.Variable(2.0)
b = tf.Variable(1.0)

with tf.GradientTape() as tape:
    y = w * 3.0 + b

gradients = tape.gradient(y, [w, b])

print("Gradient w:", gradients[0].numpy())
print("Gradient b:", gradients[1].numpy())
```
