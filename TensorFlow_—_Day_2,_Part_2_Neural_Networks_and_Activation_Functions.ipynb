##What is an activation function?
An activation function transforms a neuron's output. It allows a neural network to learn nonlinear relationships.

Here are three important activation functions:

### **ReLU**
$$f(x) = \max(0, x)$$
Returns zero for negative inputs and the input itself for positive inputs. Common in hidden layers.

### **Sigmoid**
$$f(x) = \frac{1}{1 + e^{-x}}$$
Produces values between 0 and 1. Often used for binary classification outputs.

### **Softmax**
Converts a vector of scores into probabilities that sum to 1. Often used for mutually exclusive classes.

```python
import tensorflow as tf

x = tf.constant([-3.0, -1.0, 0.0, 1.0, 3.0])

print("Input:", x.numpy())
print("ReLU:", tf.keras.activations.relu(x).numpy())
print("Sigmoid:", tf.keras.activations.sigmoid(x).numpy())

scores = tf.constant([2.0, 1.0, 0.1])
print("Softmax:", tf.keras.activations.softmax(scores).numpy())
print("Probability sum:",
      tf.reduce_sum(tf.keras.activations.softmax(scores)).numpy())
```

```python
import tensorflow as tf

# Features: [study hours, attendance percentage]
X = tf.constant([
    [1.0, 60.0],
    [2.0, 70.0],
    [3.0, 75.0],
    [4.0, 85.0],
    [5.0, 95.0],
    [2.0, 80.0],
    [4.0, 70.0],
    [5.0, 85.0]
])

# Synthetic target marks
y = tf.constant([35.0, 45.0, 55.0, 70.0,
                 85.0, 50.0, 65.0, 80.0])

model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.summary()
```

##compile and Run Model

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss="mean_squared_error",
    metrics=["mae"]
)

history = model.fit(
    X,
    y,
    epochs=200,
    batch_size=4,
    verbose=0
)

print("Final loss:", history.history["loss"][-1])
print("Final MAE:", history.history["mae"][-1])
```

##Make Prediction

```python
new_student = tf.constant([[3.0, 80.0]])

prediction = model.predict(new_student, verbose=0)

print("Predicted marks:", prediction[0][0])
```
