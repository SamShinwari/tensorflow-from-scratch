#First Tensor Program

```python
import tensorflow as tf

x = tf.constant(10)

print(x)
```

#Scalar

A single number is called a scalar.

```python
x = tf.constant(10)
print(x)
print(x.shape)
print(x.ndim)
print(x.dtype)
```

#Vector
###1-dimensional vector

```python
x = tf.constant([1, 2, 3, 4, 5])

print(x)
print(x.shape)
print(x.ndim)
```

#Matrix

```python
x = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

print(x)
print(x.shape)
print(x.ndim)
```

#3-D Tensor


```python
x = tf.constant([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print(x)
print(x.shape)
print(x.ndim)
```

#Basic Tensor Operations

```python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

c = a + b

print(c)
```

#More tensor operation

```python
a = tf.constant([10, 20, 30])
b = tf.constant([2, 5, 10])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```
