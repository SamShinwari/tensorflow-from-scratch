#2. Tensor Comparisons
TensorFlow can compare values and return Boolean tensors containing True or False.

##Example 1: Greater than

```python
x = tf.constant([10, 25, 30, 5, 40])

result = x > 20

print(result)
```

#Example 2: Equal to

```python
x = tf.constant([10, 20, 30, 20])

print(x == 20)
```

#Example 3: Find values that satisfy a condition

###Use tf.boolean_mask() to select elements where a condition is true.

```python
scores = tf.constant([45, 78, 92, 35, 85])

passed = tf.boolean_mask(scores, scores >= 50)

print(passed)
```

#3. Practical Example: Student Marks
Let's combine broadcasting, comparisons, and tensor operations in a small real-world example.
Suppose three students have marks in three subjects.

```python
import tensorflow as tf

marks = tf.constant([
    [70, 80, 90],
    [60, 75, 85],
    [40, 50, 55]
])

print("Marks:")
print(marks)

print("Average marks:", tf.reduce_mean(
    tf.cast(marks, tf.float32), axis=1
))
```

##4. Practical Example: Normalizing Values
Neural networks often work better when input features are on a comparable numerical scale.

A simple technique is min-max normalization:

```python
x = tf.constant([10.0, 20.0, 30.0, 40.0, 50.0])

x_min = tf.reduce_min(x)
x_max = tf.reduce_max(x)

x_normalized = (x - x_min) / (x_max - x_min)

print(x_normalized)
```

###Exercise 4: Normalization

Normalize [5.0, 10.0, 15.0, 20.0, 25.0] to the range [0, 1] using tf.reduce_min() and tf.reduce_max().

```python
marks = tf.constant([5.0, 10.0, 15.0, 20.0, 25.0])
print(marks)
tf_min  = tf.reduce_min(marks)
tf_max = tf.reduce_max(marks)
print(tf_min)
print(tf_max)

marks_normalized = (marks - tf_min) / (tf_max - tf_min)
print(marks_normalized)
```
