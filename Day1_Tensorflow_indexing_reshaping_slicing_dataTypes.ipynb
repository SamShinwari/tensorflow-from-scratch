# TensorFlow — Day 1, Part 2
Today we'll continue with tensor indexing, slicing, reshaping, and data types.

These are fundamental because later you'll constantly manipulate tensors containing images, text sequences, batches, and model data.

## 1. Tensor Indexing
Let's create a tensor:

````python
import tensorflow as tf

x = tf.constant([10, 20, 30, 40, 50])
print(x)
````

TensorFlow uses zero-based indexing, just like Python and NumPy.

```
Index:   0   1   2   3   4
Value:  10  20  30  40  50
```

````python
print(x[0])
print(x[1])
print(x[4])
````

## 2. Negative Indexing
You can also count from the end:

* `x[-1]` → last element
* `x[-2]` → second-last element

````python
print(x[-1])
print(x[-2])
````

## 3. Tensor Slicing

Suppose we have our 1D tensor:
`x = tf.constant([10, 20, 30, 40, 50])`

We can select multiple elements using the `[start : stop]` syntax, where the `stop` index is not included.

Let's print elements from index 1 up to (but not including) index 4:

````python
x = tf.constant([10, 20, 30, 40, 50])
print(x[1:4])
````

## 4. More Slicing

We can omit start or stop values:
* `[:3]` means start from the beginning up to index 3
* `[2:]` means start from index 2 to the end
* `[:]` means select everything

````python
print(x[:3])
print(x[2:])
print(x[:])
````

## 5. Indexing a Matrix

Now, let's create a 2D tensor (a matrix):

```
        Column
        0   1   2

Row 0  10  20  30
Row 1  40  50  60
Row 2  70  80  90
```

````python
x_matrix = tf.constant([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Access 50 (Row 1, Column 1)
print(x_matrix[1, 1])

# Access 70 (Row 2, Column 0)
print(x_matrix[2, 0])

# Access 30 (Row 0, Column 2)
print(x_matrix[0, 2])
````

## 6. Select an Entire Row

To access complete rows, specify the row index:

````python
# First row
print(x_matrix[0])

# Second row
print(x_matrix[1])
````

## 7. Select an Entire Column

We use `:` to specify 'all rows' along with a single column index:

````python
# First column
print(x_matrix[:, 0])

# Second column
print(x_matrix[:, 1])

# Third column
print(x_matrix[:, 2])
````

## 8. Reshaping Tensors

Let's reshape a 1D tensor of 6 elements into a 2x3 matrix.

````python
x_six = tf.constant([1, 2, 3, 4, 5, 6])
print("Original shape:", x_six.shape)

x_reshaped = tf.reshape(x_six, (2, 3))
print("Reshaped Tensor:\n", x_reshaped)
print("New shape:", x_reshaped.shape)
````

## 9. Another Reshape

We can reshape the same six numbers into a 3x2 matrix. Note that the total number of elements must remain identical.

````python
x_reshaped_3x2 = tf.reshape(x_six, (3, 2))
print(x_reshaped_3x2)
````

## 10. Flattening a Tensor

We can flatten a matrix using `-1`, which tells TensorFlow to automatically determine that dimension size.

````python
x_matrix_flat = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

x_flat = tf.reshape(x_matrix_flat, (-1,))
print(x_flat)
````

## 11. Data Types

TensorFlow infers types automatically (e.g., standard integers are `int32`, standard floats are `float32`).

````python
x_int = tf.constant([1, 2, 3])
print("Integer dtype:", x_int.dtype)

x_float = tf.constant([1.0, 2.0, 3.0])
print("Float dtype:", x_float.dtype)
````

## 12. Explicitly Specify the Data Type

You can override inference by passing the `dtype` parameter.

````python
x_explicit = tf.constant([1, 2, 3], dtype=tf.float32)
print(x_explicit)
print(x_explicit.dtype)
````

## 13. Why float32 Matters

Calculations like $y = wx + b$ and model optimization gradients require floating point precision.

````python
weights = tf.constant([0.25, 0.5, 0.75], dtype=tf.float32)
print(weights)
````

## 14. Converting Data Types

We can cast tensors to different types using `tf.cast()`.

````python
x_int = tf.constant([1, 2, 3])
x_casted = tf.cast(x_int, dtype=tf.float32)
print(x_casted)
print(x_casted.dtype)
````
