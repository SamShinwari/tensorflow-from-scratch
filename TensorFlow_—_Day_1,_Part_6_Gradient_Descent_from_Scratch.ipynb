# TensorFlow — Day 1, Part 6: Gradient Descent from Scratch

Today, we'll learn how a machine learning model learns from its mistakes by updating its weights.

In the previous lesson, you learned `tf.Variable()` and `tf.GradientTape()`. Now we'll combine them to implement gradient descent.

### 1. What is gradient descent?
Imagine you want to predict a student's exam marks from their study hours.

We have a simple model:
$$\hat{y} = wx + b$$

Where:
* $x$ = study hours
* $w$ = weight
* $b$ = bias
* $\hat{y}$ = predicted marks

Suppose a student studies for 2 hours, and we want to predict 4 marks in our simplified example.

Initially, our model has $w=1$ and $b=0$.
$$\hat{y} = 1(2) + 0 = 2$$

Our prediction is 2, but the target is 4. The model needs to adjust its weight.

Gradient descent helps us find parameter updates that reduce the model's loss.

### 2. Understand the loss function
For a single example, we'll use squared error:
$$L = (\hat{y} - y)^2$$

Here:
* $y$ is the correct answer.
* $\hat{y}$ is the model's prediction.
* $L$ is the loss.

For our example:
$$L = (2 - 4)^2 = 4$$

Our goal is to reduce this loss toward zero.

### 3. The gradient descent formula
The basic update rule is:
$$w_{\text{new}} = w_{\text{old}} - \alpha \frac{\partial L}{\partial w}$$

Here, $\alpha$ is the learning rate. It controls how large each update is.
* A very small learning rate can make learning slow.
* A very large learning rate can cause unstable updates.
* A suitable learning rate helps the model converge.

We use the same principle to update the bias.

#Our first gradient descent program

```python
import tensorflow as tf

# Training example
x = tf.constant(2.0)
y_true = tf.constant(4.0)

# Model parameters
w = tf.Variable(1.0)
b = tf.Variable(0.0)

# Learning rate
learning_rate = 0.1

# Calculate prediction and loss
with tf.GradientTape() as tape:
    y_pred = w * x + b
    loss = (y_pred - y_true) ** 2

# Calculate gradients
dw, db = tape.gradient(loss, [w, b])

# Update parameters
w.assign_sub(learning_rate * dw)
b.assign_sub(learning_rate * db)

print("Prediction before update: 2.0")
print("Loss before update:", loss.numpy())
print("Gradient of w:", dw.numpy())
print("Gradient of b:", db.numpy())
print("Updated w:", w.numpy())
print("Updated b:", b.numpy())
```

##Let TensorFlow train for multiple steps

```python
import tensorflow as tf

x = tf.constant(2.0)
y_true = tf.constant(4.0)

w = tf.Variable(1.0)
b = tf.Variable(0.0)

learning_rate = 0.1

for step in range(10):

    with tf.GradientTape() as tape:
        y_pred = w * x + b
        loss = (y_pred - y_true) ** 2

    dw, db = tape.gradient(loss, [w, b])

    w.assign_sub(learning_rate * dw)
    b.assign_sub(learning_rate * db)

    print(
        f"Step {step + 1}: "
        f"Loss={loss.numpy():.6f}, "
        f"w={w.numpy():.4f}, "
        f"b={b.numpy():.4f}"
    )
```

#Exercise 1: Change the learning rate

Run the single-update example with learning rates 0.01, 0.1 and 0.5. Compare the updated weight and loss.


```python
import tensorflow as tf

print("=== Exercise 1: Change the learning rate ===")
learning_rates = [0.01, 0.1, 0.5]

for lr in learning_rates:
    x = tf.constant(2.0)
    y_true = tf.constant(4.0)
    w = tf.Variable(1.0)
    b = tf.Variable(0.0)

    with tf.GradientTape() as tape:
        y_pred = w * x + b
        loss = (y_pred - y_true) ** 2

    dw, db = tape.gradient(loss, [w, b])
    w.assign_sub(lr * dw)
    b.assign_sub(lr * db)

    print(f"Learning Rate: {lr:<5} | Updated w: {w.numpy():.4f} | Updated b: {b.numpy():.4f}")

```

#Exercise 2: Change the target

Set x = 3.0 and y_true = 6.0. Start with w = 1.0 and b = 0.0. Perform five gradient descent steps.

```python
x = tf.constant(3.0)
y_true = tf.constant(6.0)
w = tf.Variable(1.0)
b = tf.Variable(0.0)
learning_rate = 0.1

for step in range(5):
    with tf.GradientTape() as tape:
        y_pred = w * x + b
        loss = (y_pred - y_true) ** 2
    dw, db = tape.gradient(loss, [w, b])
    w.assign_sub(learning_rate * dw)
    b.assign_sub(learning_rate * db)
    print(f"Step {step + 1}: Loss={loss.numpy():.6f}, w={w.numpy():.4f}, b={b.numpy():.4f}")
```

#Exercise 3: Track the loss

Train for 20 steps and store each step's loss in a Python list. Print the list to see how the loss changes.

```python

x = tf.constant(2.0)
y_true = tf.constant(4.0)
w = tf.Variable(1.0)
b = tf.Variable(0.0)
learning_rate = 0.05

losses = []
for step in range(20):
    with tf.GradientTape() as tape:
        y_pred = w * x + b
        loss = (y_pred - y_true) ** 2
    losses.append(loss.numpy())

    dw, db = tape.gradient(loss, [w, b])
    w.assign_sub(learning_rate * dw)
    b.assign_sub(learning_rate * db)

print("Losses tracked over 20 steps:")
for i, loss_val in enumerate(losses):
    print(f"Step {i+1:02d}: {loss_val:.6f}")
```
