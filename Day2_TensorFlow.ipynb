### 1. Importing TensorFlow
In this cell, we import the TensorFlow library so that we can use its API to create and manipulate tensors.

```python
import tensorflow as tf
```

### 2. Creating a 0-Dimensional Tensor (Scalar)
Here, we use `tf.constant` to define a scalar tensor (rank 0 tensor). It holds a single integer value: `7`.

```python
scalar  = tf.constant(7)
scalar
```

### 3. Creating a 1-Dimensional Tensor (Vector)
This cell creates a 1D tensor, also known as a vector. It contains two elements: `[10, 10]`.

```python
vector = tf.constant([10,10])
vector
```

### 4. Creating a 2-Dimensional Tensor (Matrix)
Here, we create a 2D tensor (a matrix) with 2 rows and 2 columns. Its shape is `(2, 2)`.

```python
matrix = tf.constant([[10,7],[7,10]])
matrix
```

### 5. Creating a 3-Dimensional Tensor
In this cell, we define a 3D tensor of shape `(2, 2, 2)`. We then inspect its properties:
* **Shape**: dimensions of the tensor (`2, 2, 2`)
* **ndim**: the number of dimensions/axes (`3`)

```python
tensor = tf.constant([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print(tensor)
print(tensor.shape)
print(tensor.ndim)
```

#Tensorflow VS Numpy

#### A. NumPy Array representation
This cell initializes a standard 1D array using NumPy.

```python
import numpy as np

x = np.array([1, 2, 3])
print(x)
```

#### B. TensorFlow equivalent
This cell creates the equivalent 1D tensor in TensorFlow using `tf.constant`.

```python
import tensorflow as tf

x = tf.constant([1, 2, 3])
print(x)
```

### 6. Element-wise Addition
We create two vectors `a` and `b`, and then perform an element-wise addition (`a + b`).

```python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

c = a + b

print(c)
```

### 7. Core Math Operations
This cell demonstrates basic arithmetic operations on two tensors of identical shape: addition (`+`), subtraction (`-`), multiplication (`*`), and division (`/`).

```python
a = tf.constant([10, 20, 30])
b = tf.constant([2, 5, 10])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

### 8. Tensor Properties and Operations
Similar to the previous cell, we run arithmetic operations on 5-element tensors and output critical metadata attributes:
* **`.dtype`**: The data type of elements inside the tensor.
* **`.ndim`**: The rank/number of dimensions.
* **`.shape`**: The shape description of the tensor.

```python
t1 = tf.constant( [10, 20, 30, 40, 50] )
t2 = tf.constant([2,4,5,8,10])

print(t1+t2)
print(t1-t2)
print(t1*t2)
print(t1/t2)
print(t1.dtype)
print(t1.ndim)
print(t1.shape)
```

### 9. 2D Tensor Matrix Evaluation
Here, we define a 3x3 matrix and examine its structural properties like shape, datatype, and dimensions.

```python
arr = tf.constant([[1, 2, 3],
 [4, 5, 6],
 [7, 8, 9]])
print(arr)
print(arr.shape)
print(arr.dtype)
print(arr.ndim)
```
